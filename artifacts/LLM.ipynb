```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from transformers.utils import logging

logging.set_verbosity_info()

model_name = "Qwen/Qwen2.5-1.5B-Instruct"

print("Step 1: Loading tokenizer...", flush=True)

tokenizer = AutoTokenizer.from_pretrained(
    model_name,
    local_files_only=True
)

print("Step 2: Loading model onto GPU...", flush=True)

llm = AutoModelForCausalLM.from_pretrained(
    model_name,
    dtype=torch.float16,
    device_map="cuda",
    local_files_only=True,
    low_cpu_mem_usage=True
)

llm.eval()

print("\nModel loaded successfully!", flush=True)
print("Model device:", llm.device)
print("GPU:", torch.cuda.get_device_name(0))
print("GPU memory allocated:",
      f"{torch.cuda.memory_allocated() / 1024**3:.2f} GB")
```

```python
# Define a simple test conversation
messages = [
    {
        "role": "system",
        "content": "You are a helpful research assistant. Answer clearly and concisely."
    },
    {
        "role": "user",
        "content": "Explain retrieval-augmented generation in simple terms."
    }
]

# Convert the conversation into the model's expected format
prompt = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

# Tokenize the prompt and move it to the GPU
inputs = tokenizer(prompt, return_tensors="pt").to(llm.device)

# Generate a response
with torch.inference_mode():
    outputs = llm.generate(
        **inputs,
        max_new_tokens=200,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

# Extract only the newly generated tokens
new_tokens = outputs[0][inputs["input_ids"].shape[1]:]

# Convert the tokens into readable text
answer = tokenizer.decode(new_tokens, skip_special_tokens=True)

print("Question: Explain retrieval-augmented generation in simple terms.\n")
print("Answer:\n", answer)
```
